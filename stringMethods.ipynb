# expandTabs() Methods

```python
x = "M\tu\tm\tb\ta\ti"

print(x)
print(x.expandtabs())
print(x.expandtabs(2))
print(x.expandtabs(4))
print(x.expandtabs(8))
```

# format Method

```python
txt1 = "My name is {fname}, my age is {age}".format(fname ="Rishbh" ,age=100)
txt2 = "My name is {0}, my age is {1}".format("Rishbh" ,100)
txt3 = "My name is {}, my age is {}".format("Rishbh" ,100)

print(txt1)
print(txt2)
print(txt3)
```

# format_map() Method

```python
loop = {"Name" : "Lisa","age" :15}
txt3 ="Happy Birthday {Name} you are now on level {age}!"


print(txt3.format_map(loop))
```
