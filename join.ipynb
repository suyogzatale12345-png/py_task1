# join() Method

```python
md = {"name": "rahul", "country": "india"}
Sp = "Subject"

x = Sp.join(md)

print(x)
```

```python
md = {"name", "rahul", "country", "india"}
Sp = "Subject"

x = Sp.join(md)

print(x)
```

```python
md = {"name": "rahul", "country" : "india"}
Sp = "Subject"

x = Sp.join(md)

print(x)
```

```python
md = {"Song": "har kisi ko", "country": "india"}
Sp = "har kisi ko"

x = Sp.join(md)

print(x)
```
